```sql
CREATE SCHEMA IF NOT EXISTS emissions.bronze;
CREATE SCHEMA IF NOT EXISTS emissions.silver;
CREATE SCHEMA IF NOT EXISTS emissions.gold;


SELECT *
FROM read_files(
    '/Volumes/emissions/bronze/raw_bronze/Emissions_Data_2023.csv',
    format => 'csv',
    header => true
)
LIMIT 10;

-------------------------------------------------------------------------------------------------------------------
-- Creating Bronze Delta table while preserving original source column names

CREATE OR REPLACE TABLE emissions.bronze.emissions_raw
USING DELTA
TBLPROPERTIES (
    'delta.columnMapping.mode' = 'name'
)
AS

SELECT
    *,
    current_timestamp() AS ingestion_timestamp,
    _metadata.file_name AS source_file

FROM read_files(
    '/Volumes/emissions/bronze/raw_bronze/Emissions_Data_2023.csv',
    format => 'csv',
    header => true
);


--------------------------------------------------------------------------

SELECT COUNT(*) AS bronze_row_count
FROM emissions.bronze.emissions_raw;

SELECT *
FROM emissions.bronze.emissions_raw
LIMIT 10;

SELECT
    source_file,
    COUNT(*) AS records
FROM emissions.bronze.emissions_raw
GROUP BY source_file;


```
