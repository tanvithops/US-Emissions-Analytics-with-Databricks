```sql
-- QUERY 1: Retrieving latitude, longitude, and GHG emissions for geographic mapping
SELECT  
    latitude, 
    longitude, 
    `GHG emissions mtons CO2e` 
FROM emissions_data; 

-----------------------------------------------------------

-- QUERY 2: Calculating emissions per person for each county and showing the top 10 highest values
SELECT 
    county_state_name, 
    population, 
    CAST( 
        REPLACE(`GHG emissions mtons CO2e`, ',', '') 
        AS DOUBLE 
    ) / 
    NULLIF( 
        CAST( 
            REPLACE(population, ',', '') 
            AS DOUBLE 
        ), 
        0 
    ) AS emissions_per_person 
FROM emissions_data 
ORDER BY emissions_per_person DESC 
LIMIT 10; 

-----------------------------------------------------------------

-- QUERY 3: Calculating total GHG emissions for each state and ranking them from highest to lowest
SELECT 
    state_abbr, 
    SUM(
        CAST( 
            REPLACE(`GHG emissions mtons CO2e`, ',', '') 
            AS DOUBLE 
        )
    ) AS Total_Emissions 
FROM emissions_data 
GROUP BY state_abbr 
ORDER BY Total_Emissions DESC;
-- LIMIT 10;

----------------------------------------------------------------------

-- QUERY 4: Calculating total emissions from the top 10 states and their percentage of overall U.S. emissions
WITH top10 AS ( 
    SELECT 
        state_abbr, 
        SUM( 
            CAST( 
                REPLACE(`GHG emissions mtons CO2e`, ',', '') 
                AS DOUBLE 
            ) 
        ) AS Total_Emissions 
    FROM emissions_data 
    GROUP BY state_abbr 
    ORDER BY Total_Emissions DESC 
    LIMIT 10 
) 

SELECT 
    SUM(Total_Emissions) AS Top10_Emissions, 
    ( 
        SUM(Total_Emissions) / 
        ( 
            SELECT SUM( 
                CAST( 
                    REPLACE(`GHG emissions mtons CO2e`, ',', '') 
                    AS DOUBLE 
                ) 
            ) 
            FROM emissions_data 
        ) 
    ) * 100 AS Top10_Percentage 
FROM top10; 

----------------------------------------------------------------------------

-- QUERY 5: Identifying the top 10 counties with the highest total GHG emissions
SELECT 
    county_state_name, 
    population, 
    CAST( 
        REPLACE(`GHG emissions mtons CO2e`, ',', '') 
        AS DOUBLE 
    ) AS Total_Emissions 
FROM emissions_data 
ORDER BY Total_Emissions DESC 
LIMIT 10;
```
